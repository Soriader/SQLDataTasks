```python
%sql
CREATE CATALOG IF NOT EXISTS prod;

CREATE SCHEMA IF NOT EXISTS prod.sales;


GRANT USE CATALOG
ON CATALOG prod
TO `analysts`;

GRANT USE SCHEMA, SELECT
ON SCHEMA prod.sales
TO `analysts`;
```

```python
REVOKE SELECT
ON SCHEMA prod.sales
FROM `analysts`;
```
