```python
import pandas as pd
import numpy as np

df = pd.read_csv("sample_orders.csv")

country_arr = df["country"].to_numpy()

cats, inv = np.unique(country_arr, return_inverse=True)
one_hot = np.eye(len(cats), dtype=int)[inv]

print("=== One-hot country ===")
print("cats:")
print(cats)

print("one_hot:")
print(one_hot)

quantity_arr = df["quantity"].to_numpy()
price_arr = df["price_each"].to_numpy()

corr_quantity_price = np.corrcoef(quantity_arr, price_arr)[0, 1]

print("=== Correlation quantity vs price_each ===")
print(corr_quantity_price)
print(round(corr_quantity_price, 2))

quartiles = np.percentile(price_arr, [25, 50, 75])
price_bucket = np.digitize(price_arr, quartiles, right=True) + 1

print("=== Quartiles ===")
print(quartiles)

print("=== Price buckets ===")
print(price_bucket)

df["price_bucket"] = price_bucket

print(df[["price_each", "price_bucket"]].head())
```
