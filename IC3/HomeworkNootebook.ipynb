```python
import pandas as pd
import numpy as nb

df = pd.read_csv('sample_orders.csv', parse_dates=['placed_at'])

is_paid = df.query("status == 'paid'")

revenue_by_country = ((is_paid.assign(revenue =(lambda r: r.price_each * r.quantity))
                      .groupby(['country'])
                      .agg(revenue = ("price_each", "sum"), orders=("id", "count")))
                      .sort_values('revenue', ascending=False))

print(revenue_by_country)

revenue_by_country2 = (
    df
    .query("status == 'paid'")
    .assign(revenue=lambda r: r["price_each"] * r["quantity"])
    .groupby("country", as_index=False)
    .agg(
        revenue=("revenue", "sum"),
        orders=("id", "count")
    )
    .sort_values("revenue", ascending=False)
)

print(revenue_by_country2)
```
