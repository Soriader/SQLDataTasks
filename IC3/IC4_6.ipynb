```python
import pandas as pd
import polars as pl
from time import perf_counter

start = perf_counter()
df_pandas = pd.read_csv('orders.csv')
df_pl = df_pandas[df_pandas["country"] == "PL"]
df_pl_quantity = df_pl["quantity"].sum()
end = perf_counter()
print(end - start)
#time: 0.9217010130000745 result for sum = 150811

start = perf_counter()
df_polars = pl.scan_csv("orders.csv")
df_pl_polars = df_polars.filter(pl.col("country") == "PL")
df_pl_quantity_polars = df_pl_polars.select(pl.col("quantity").sum()).collect()
end = perf_counter()
print(end - start)
#time: 0.1875481640001908 result for sum = 150811
```
